## Opdracht 3: Kalibratie sensor en foto

```python
# verwerk hier jullie kalibratiemetingen
from scipy.optimize import curve_fit

afstand = np.array([7.1,10.0,13.0,16.0,19.0,21.0,24.0,30.0,35.0,40.0,45.0,50.0,55.0,60.0,65.0,70.0]) #cm
output = np.array([205,285,372,458,541,615,710,879,1027,1170,1315,1455,1600,1755,1895,2025]) 

def function(x,a,b):
    return a*x+b

popt, pcov = curve_fit(function, output, afstand)
a=popt[0]
b=popt[1]

print(a,b)
x_fit = np.linspace(output.min(), output.max(), 500)

plt.figure()
plt.plot(output,afstand,'r.',label='kalibratiemeting')
plt.plot(x_fit, function(x_fit,a,b), 'b-', label='fit kalibratie')
plt.xlabel("output (bits)")
plt.ylabel("afstand (cm)")
plt.grid()
plt.legend()
plt.show()
```
